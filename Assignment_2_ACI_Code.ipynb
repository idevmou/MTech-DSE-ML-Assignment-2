## Data Preprocessing

### Separate features (X) and target (y)

```python
X = df.drop(columns=[target_column])
y = df[target_column]

print("Features shape:", X.shape)
print("Target shape:", y.shape)

print("\nFeature columns:")
print(X.columns.tolist())
```

### Encode target labels

```python
from sklearn.preprocessing import LabelEncoder

label_encoder = LabelEncoder()

y_encoded = label_encoder.fit_transform(y)

print("Original Classes:")
print(label_encoder.classes_)

print("\nEncoded Classes:")
print(np.unique(y_encoded))
```

### Split training and testing data

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y_encoded,
    test_size=0.20,
    random_state=42,
    stratify=y_encoded
)

print("Training Features:", X_train.shape)
print("Testing Features:", X_test.shape)

print("Training Target:", y_train.shape)
print("Testing Target:", y_test.shape)
```

Why random_state=42?

It ensures reproducibility. Running the split again with the same data and settings produces the same partition.

Why stratify=y_encoded?

It preserves approximately the same class proportions in training and testing datasets.
3.5 Feature scaling
Some features have different numerical ranges.

For example:
- Area might be in thousands.
- Eccentricity might be between 0 and 1.
Algorithms such as KNN, Logistic Regression and SVM can be affected by these differences.

We use StandardScaler.
\[
z = \frac{x-\mu}{\sigma}
\]

Where:
- \(x\) = Original value
- \(\mu\) = Training mean
- \(\sigma\) = Training standard deviation
Execute:

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

# Fit only on training data
X_train_scaled = scaler.fit_transform(X_train)

# Transform test data using training statistics
X_test_scaled = scaler.transform(X_test)

print("Original training shape:", X_train.shape)
print("Scaled training shape:", X_train_scaled.shape)

print("\nOriginal testing shape:", X_test.shape)
print("Scaled testing shape:", X_test_scaled.shape)
```

### Verify preprocessing

```python
print("Training dataset size:", len(X_train))
print("Testing dataset size:", len(X_test))

print("\nMissing values in training:", X_train.isnull().sum().sum())
print("Missing values in testing:", X_test.isnull().sum().sum())

print("\nTraining class distribution:")
print(pd.Series(y_train).value_counts().sort_index())

print("\nTesting class distribution:")
print(pd.Series(y_test).value_counts().sort_index())
```

### Model 1 – Logistic Regression

```python
from sklearn.linear_model import LogisticRegression

# Create model
logistic_model = LogisticRegression(
    max_iter=2000,
    random_state=42
)

# Train model
logistic_model.fit(X_train_scaled, y_train)

# Predict test data
y_pred_lr = logistic_model.predict(X_test_scaled)

# Predict probabilities for AUC
y_prob_lr = logistic_model.predict_proba(X_test_scaled)

print("Logistic Regression training completed!")
print("Predictions:", y_pred_lr[:10])
```

### Evaluate Logistic Regression

```python
from sklearn.metrics import (
    accuracy_score,
    roc_auc_score,
    precision_score,
    recall_score,
    f1_score,
    matthews_corrcoef
)

lr_accuracy = accuracy_score(y_test, y_pred_lr)

lr_auc = roc_auc_score(
    y_test,
    y_prob_lr,
    multi_class="ovr",
    average="weighted"
)

lr_precision = precision_score(
    y_test, y_pred_lr, average="weighted", zero_division=0
)

lr_recall = recall_score(
    y_test, y_pred_lr, average="weighted", zero_division=0
)

lr_f1 = f1_score(
    y_test, y_pred_lr, average="weighted", zero_division=0
)

lr_mcc = matthews_corrcoef(y_test, y_pred_lr)

print("LOGISTIC REGRESSION RESULTS")
print("---------------------------")
print("Accuracy :", round(lr_accuracy, 4))
print("AUC      :", round(lr_auc, 4))
print("Precision:", round(lr_precision, 4))
print("Recall   :", round(lr_recall, 4))
print("F1 Score :", round(lr_f1, 4))
print("MCC      :", round(lr_mcc, 4))
```

### Understand the six evaluation metrics

#### Metric	What does it measure?
Accuracy	Overall correct predictions
AUC	Ability to distinguish between classes
Precision	Correctness of positive predictions
Recall	Ability to identify actual positive cases
F1 Score	Balance between precision and recall
MCC	Overall correlation between actual and predicted classes

For our multiclass dataset, we're using weighted averages for precision, recall, F1 and AUC.

### <b>Conclusion Logistic Regression:-</b>
Logistic Regression achieved 92.07% accuracy and 99.34% weighted AUC. The model demonstrated strong classification performance, with balanced precision and recall. Its MCC score of 0.9042 indicates strong agreement between actual and predicted classifications.

## Support Vector Machine (SVM)

SVM stands for Support Vector Machine.

It tries to find the best decision boundary (hyperplane) that separates different classes.

Its objective is to maximize the margin between the decision boundary and the nearest observations.

Important terms:
- Hyperplane: Decision boundary separating classes.
- Margin: Distance between the boundary and nearest observations.
- Support Vectors: Observations closest to the decision boundary.
- Kernel: Function that helps model nonlinear decision boundaries.
Why are we using RBF kernel?
Our dataset may contain nonlinear relationships between bean measurements.

The RBF kernel allows SVM to learn nonlinear decision boundaries.

### Implement

```python
from sklearn.svm import SVC

# Create SVM model
svm_model = SVC(
    kernel="rbf",
    C=1.0,
    probability=True,
    random_state=42
)

# Train the model
svm_model.fit(X_train_scaled, y_train)

# Predict test data
y_pred_svm = svm_model.predict(X_test_scaled)

# Predict class probabilities
y_prob_svm = svm_model.predict_proba(X_test_scaled)

print("SVM training completed!")
print("Predictions:", y_pred_svm[:10])
```

### Evaluate

```python
svm_accuracy = accuracy_score(y_test, y_pred_svm)

svm_auc = roc_auc_score(
    y_test,
    y_prob_svm,
    multi_class="ovr",
    average="weighted"
)

svm_precision = precision_score(
    y_test, y_pred_svm,
    average="weighted",
    zero_division=0
)

svm_recall = recall_score(
    y_test, y_pred_svm,
    average="weighted",
    zero_division=0
)

svm_f1 = f1_score(
    y_test, y_pred_svm,
    average="weighted",
    zero_division=0
)

svm_mcc = matthews_corrcoef(y_test, y_pred_svm)

print("SUPPORT VECTOR MACHINE RESULTS")
print("-------------------------------")
print("Accuracy :", round(svm_accuracy, 4))
print("AUC      :", round(svm_auc, 4))
print("Precision:", round(svm_precision, 4))
print("Recall   :", round(svm_recall, 4))
print("F1 Score :", round(svm_f1, 4))
print("MCC      :", round(svm_mcc, 4))
```
